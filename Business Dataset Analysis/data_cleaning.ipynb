```python

import pandas as pd
import numpy as np
from pathlib import Path

# Define project paths
BASE_DIR = Path.cwd()

INPUT_FILE = BASE_DIR / "india_startup_funding_sample.csv"
OUTPUT_DIR = BASE_DIR / "outputs"

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

# Step 1: Load original dataset
df = pd.read_csv(INPUT_FILE)

print("Original dataset shape:", df.shape)

# Step 2: Remove unwanted spaces from text columns
text_columns = df.select_dtypes(include="object").columns

for column in text_columns:
    df[column] = df[column].str.strip()

# Step 3: Convert announced_date into datetime
df["announced_date"] = pd.to_datetime(
    df["announced_date"],
    errors="coerce"
)

# Step 4: Check duplicate funding round IDs
print(
    "Duplicate round IDs:",
    df["round_id"].duplicated().sum()
)

# Step 5: Handle missing categorical values
categorical_columns = [
    "city",
    "sector",
    "lead_investor"
]

for column in categorical_columns:
    df[column] = df[column].fillna("Unknown")

# Step 6: Handle missing news information
news_columns = ["news_source", "news_excerpt"]

for column in news_columns:
    df[column] = df[column].fillna("Unavailable")

# Step 7: Keep missing funding amounts unchanged
# Missing amounts represent undisclosed funding,
# not necessarily zero funding.

# Step 8: Check data consistency
print("\nYear mismatches:")
print((df["announced_date"].dt.year != df["year"]).sum())

print("\nInvalid founding years:")
print((df["founded_year"] > df["year"]).sum())

# Step 9: Generate missing-value report
missing_report = df.isnull().sum()

print("\nMissing values after cleaning:")
print(missing_report)

# Step 10: Save cleaned dataset
output_file = OUTPUT_DIR / "indian_startup_funding_cleaned.csv"

df.to_csv(output_file, index=False)

print("\nCleaning completed successfully!")
print("Cleaned dataset shape:", df.shape)
print("Saved to:", output_file)
```
