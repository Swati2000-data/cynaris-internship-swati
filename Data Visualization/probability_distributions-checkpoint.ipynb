```python
import pandas as pd

# Create contingency table
data = {
    "Purchased": [40, 20],
    "Not Purchased": [60, 80]
}

df = pd.DataFrame(
    data,
    index=["Premium", "Standard"]
)

# Calculate totals
df["Total"] = df.sum(axis=1)
df.loc["Total"] = df.sum()

print("Contingency Table:")
print(df)

# Total customers
total_customers = df.loc["Total", "Total"]

# P(A): Probability of Premium customer
P_A = df.loc["Premium", "Total"] / total_customers

# P(A intersection B): Premium and purchased
P_A_and_B = df.loc["Premium", "Purchased"] / total_customers

# P(A|B): Premium given that customer purchased
P_A_given_B = (
    df.loc["Premium", "Purchased"] /
    df.loc["Total", "Purchased"]
)

print("\nProbability Results:")
print("P(A):", round(P_A, 3))
print("P(A intersection B):", round(P_A_and_B, 3))
print("P(A|B):", round(P_A_given_B, 3))
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Define parameters
mean = 100
std_dev = 15

# Generate values
x = np.linspace(mean - 4 * std_dev,
                mean + 4 * std_dev, 1000)

# Calculate normal distribution
y = norm.pdf(x, mean, std_dev)

# Plot distribution
plt.figure(figsize=(10, 5))

plt.plot(x, y, label="Normal Distribution")

# Mark mean and standard deviations
for i in range(-3, 4):
    value = mean + i * std_dev
    plt.axvline(value, linestyle="--",
                label=f"{i} SD" if i != 0 else "Mean")

plt.title("Normal Distribution")
plt.xlabel("Employee Performance Score")
plt.ylabel("Probability Density")
plt.legend()
plt.grid(alpha=0.3)

plt.show()

# Print empirical rule
print("68% of observations:",
      mean - std_dev, "to", mean + std_dev)

print("95% of observations:",
      mean - 2 * std_dev, "to", mean + 2 * std_dev)

print("99.7% of observations:",
      mean - 3 * std_dev, "to", mean + 3 * std_dev)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Define parameters
n = 10
p = 0.3

# Simulate 10000 experiments
np.random.seed(42)

results = np.random.binomial(
    n=n,
    p=p,
    size=10000
)

# Calculate frequency of successes
values, counts = np.unique(
    results,
    return_counts=True
)

# Plot binomial distribution
plt.figure(figsize=(9, 5))

plt.bar(values, counts / 10000)

plt.title("Binomial Distribution Simulation")
plt.xlabel("Number of Successful Purchases")
plt.ylabel("Probability")

plt.xticks(range(n + 1))
plt.grid(axis="y", alpha=0.3)

plt.show()

# Display results
print("Average number of purchases:",
      np.mean(results))

print("Expected number of purchases:",
      n * p)

print("Maximum purchases:",
      np.max(results))

print("Minimum purchases:",
      np.min(results))
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Set random seed
np.random.seed(42)

# Generate skewed population
population = np.random.exponential(
    scale=10,
    size=100000
)

# Simulation parameters
sample_size = 30
number_of_samples = 1000

# Store sample means
sample_means = []

# Generate random samples
for i in range(number_of_samples):

    sample = np.random.choice(
        population,
        size=sample_size,
        replace=True
    )

    sample_means.append(np.mean(sample))

sample_means = np.array(sample_means)

# Plot population distribution
plt.figure(figsize=(9, 5))
plt.hist(population, bins=50, density=True)

plt.title("Original Population Distribution")
plt.xlabel("Values")
plt.ylabel("Density")

plt.show()

# Plot sample means distribution
plt.figure(figsize=(9, 5))
plt.hist(sample_means, bins=30, density=True)

plt.title("Central Limit Theorem - Sample Means")
plt.xlabel("Sample Means")
plt.ylabel("Density")

plt.show()

# Compare means
print("Population Mean:",
      round(np.mean(population), 2))

print("Average of Sample Means:",
      round(np.mean(sample_means), 2))

print("Standard Error:",
      round(np.std(sample_means, ddof=1), 2))
```
