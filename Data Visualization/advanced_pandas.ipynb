```python

import pandas as pd

# 1. Create the sales DataFrame
sales = pd.DataFrame({
    "Order_ID": [101, 102, 103, 104, 105],
    "Region": ["South", "North", "South", "West", "East"],
    "Product": ["Laptop", "Mobile", "Mobile", "Laptop", "Tablet"],
    "Quantity": [2, 3, 5, 1, 4],
    "Unit_Price": [50000, 15000, 15000, 50000, 20000]
})

# Calculate total sales
sales["Sales"] = sales["Quantity"] * sales["Unit_Price"]

# 2. Create a customer DataFrame
customers = pd.DataFrame({
    "Order_ID": [101, 102, 103, 106],
    "Customer": ["Ananya", "Rahul", "Priya", "Arjun"]
})

print("SALES DATA")
print(sales)

print("\nCUSTOMER DATA")
print(customers)


# 3. Demonstrate all four join types
inner_join = pd.merge(
    sales, customers, on="Order_ID", how="inner"
)
print("\nINNER JOIN")
print(inner_join)

left_join = pd.merge(
    sales, customers, on="Order_ID", how="left"
)
print("\nLEFT JOIN")
print(left_join)

right_join = pd.merge(
    sales, customers, on="Order_ID", how="right"
)
print("\nRIGHT JOIN")
print(right_join)

outer_join = pd.merge(
    sales, customers, on="Order_ID", how="outer"
)
print("\nOUTER JOIN")
print(outer_join)


# 4. Create a pivot table: sales by region and product
pivot = pd.pivot_table(
    sales,
    values="Sales",
    index="Region",
    columns="Product",
    aggfunc="sum",
    fill_value=0
)

print("\nSALES PIVOT TABLE")
print(pivot)


# 5. Reshape wide data into long format using melt
wide_sales = pd.DataFrame({
    "Region": ["South", "North", "West"],
    "Laptop": [100000, 50000, 50000],
    "Mobile": [75000, 45000, 30000],
    "Tablet": [20000, 40000, 25000]
})

long_sales = pd.melt(
    wide_sales,
    id_vars=["Region"],
    var_name="Product",
    value_name="Sales"
)

print("\nLONG-FORMAT SALES DATA")
print(long_sales)


# 6. Chain multiple Pandas operations
summary = (
    sales
    .assign(Sales=lambda df: df["Quantity"] * df["Unit_Price"])
    .groupby("Region", as_index=False)
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Quantity=("Quantity", "sum"),
        Number_of_Orders=("Order_ID", "nunique")
    )
    .sort_values("Total_Sales", ascending=False)
    .reset_index(drop=True)
)

print("\nREGIONAL SALES SUMMARY")
print(summary)
```
