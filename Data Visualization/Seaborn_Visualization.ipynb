```python

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Create a folder for saving charts
os.makedirs("charts", exist_ok=True)

# Apply a professional Seaborn theme
sns.set_theme(style="whitegrid", palette="deep")

# Sample dataset
df = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [120, 150, 135, 180, 210, 240],
    "Advertising": [20, 25, 22, 30, 35, 40],
    "Profit": [25, 32, 28, 40, 48, 55]
})

# 1. Line Chart
plt.figure(figsize=(8, 5))
plt.plot(df["Month"], df["Sales"], marker="o", linewidth=2)
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.annotate(
    f"Highest: {df['Sales'].max()}",
    xy=(5, df["Sales"].max()),
    xytext=(3.5, 245),
    arrowprops=dict(arrowstyle="->")
)
plt.tight_layout()
plt.savefig("charts/line_chart.png", dpi=300, bbox_inches="tight")
plt.show()

# 2. Bar Chart
plt.figure(figsize=(8, 5))
bars = plt.bar(df["Month"], df["Sales"])
plt.title("Monthly Sales Comparison")
plt.xlabel("Month")
plt.ylabel("Sales")

for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height() + 3,
        str(int(bar.get_height())),
        ha="center"
    )

plt.tight_layout()
plt.savefig("charts/bar_chart.png", dpi=300, bbox_inches="tight")
plt.show()

# 3. Scatter Plot
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="Advertising", y="Sales", s=100)
plt.title("Advertising vs Sales")
plt.xlabel("Advertising Spend")
plt.ylabel("Sales")
plt.tight_layout()
plt.savefig("charts/scatter_plot.png", dpi=300, bbox_inches="tight")
plt.show()

# 4. Histogram
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="Sales", bins=5, kde=True)
plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.tight_layout()
plt.savefig("charts/histogram.png", dpi=300, bbox_inches="tight")
plt.show()

# 5. Box Plot
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, y="Sales")
plt.title("Sales Distribution and Outliers")
plt.ylabel("Sales")
plt.tight_layout()
plt.savefig("charts/box_plot.png", dpi=300, bbox_inches="tight")
plt.show()

print("All five charts saved successfully in the charts folder.")
```

Line chart: Sales increase overall from January to June.



Bar chart: June has the highest sales, at 240, while January has the lowest, at 120.



Scatter plot: Advertising spend and sales show a positive relationship in this sample dataset. This does not establish causation.



Histogram: The chart shows how the six sales observations are distributed across value ranges.



Box plot: The chart summarises the median, quartiles, and potential outliers. With only six observations, conclusions about outliers are limited.
