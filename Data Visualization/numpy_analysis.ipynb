```python
import numpy as np
import time

# 1D array
arr_1d = np.array([10, 20, 30, 40, 50])

# 2D array: rows and columns
arr_2d = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

# 3D array: multiple 2D arrays
arr_3d = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

print("1D Array:", arr_1d)
print("1D Shape:", arr_1d.shape)

print("\n2D Array:\n", arr_2d)
print("2D Shape:", arr_2d.shape)

print("\n3D Array:\n", arr_3d)
print("3D Shape:", arr_3d.shape)
```

```python
sales = np.array([
    [100, 200, 300],
    [150, 250, 350],
    [200, 300, 400]
])

print("Add 10:", sales + 10)
print("Multiply by 2:\n", sales * 2)

print("Total sales:", np.sum(sales))
print("Row totals:", np.sum(sales, axis=1))
print("Column totals:", np.sum(sales, axis=0))
```

```python
sales = np.array([120, 450, 180, 700, 320, 900])

# Boolean masking
high_sales = sales[sales > 400]
print("Sales above 400:", high_sales)

# Fancy indexing: select specific positions
selected_sales = sales[[0, 3, 5]]
print("Selected sales:", selected_sales)

# Replace values meeting a condition
sales[sales < 200] = 200
print("Updated sales:", sales)
```

```python
stock_returns = np.array([2.5, -1.0, 3.2, 1.8, -0.5, 4.0])

print("Mean:", np.mean(stock_returns))
print("Standard deviation:", np.std(stock_returns))
print("25th percentile:", np.percentile(stock_returns, 25))
print("Median:", np.median(stock_returns))

# Correlation between two variables
stock_prices = np.array([100, 105, 110, 115, 120])
trading_volume = np.array([2000, 2200, 2500, 2700, 3000])

correlation = np.corrcoef(stock_prices, trading_volume)[0, 1]
print("Correlation:", correlation)
```

```python
# Create one million values
n = 1_000_000
data = np.arange(n, dtype=np.float64)

# Python loop
start = time.perf_counter()

loop_result = []
for value in data:
    loop_result.append(value * 2)

loop_time = time.perf_counter() - start

# NumPy vectorised operation
start = time.perf_counter()

numpy_result = data * 2

numpy_time = time.perf_counter() - start

print("\nPerformance Comparison")
print("Dataset size:", n)
print(f"Python loop time: {loop_time:.6f} seconds")
print(f"NumPy time: {numpy_time:.6f} seconds")

if numpy_time > 0:
    print(f"Speedup: {loop_time / numpy_time:.2f}x")

# Verify that both methods give the same result
print("Results match:", np.allclose(loop_result, numpy_result))
```
